# RSS feed scraper

Pull the NDTV top-stories RSS feed with `feedparser` and turn the entries into a JSON structure.

## Setup

Install the two libraries if you don't already have them (uncomment the line below).

```python
# !pip install requests feedparser
```

```python
import requests
import feedparser
import json
```

## Fetch and parse the feed

```python
url = "https://feeds.feedburner.com/ndtvnews-top-stories"
feed = feedparser.parse(url)
```

## Inspect what came back

```python
print(feed.keys())
```

```python
print(feed['entries'][0])
```

## Build a JSON structure from the entries

```python
feed_data = {"articles": []}
for entry in feed["entries"]:
    article = {
        "title": entry.get('title', ''),
        "link": entry.get('link', ''),
        "description": entry.get('summary', ''),
        "detail": entry.content[0].get('value', '')
    }
    feed_data["articles"].append(article)
```

```python
print(json.dumps(feed_data, indent=4))
```
